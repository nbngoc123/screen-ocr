# Đánh giá Tập dữ liệu Synthetic & ICDAR Recognition
Trong notebook này, chúng ta sẽ trực quan hóa các ảnh chữ (text crops) đã được tự động sinh ra (Synthetic) cũng như các ảnh cắt sẵn thực tế từ tập ICDAR Recognition.

```python
import cv2
import json
import random
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# Hỗ trợ hiển thị tiếng Việt trên đồ thị
plt.rcParams['font.sans-serif'] = ['Segoe UI', 'Arial']

def show_random_samples(jsonl_path, num_samples=6):
    p = Path(jsonl_path)
    if not p.exists():
        print(f"Không tìm thấy file {jsonl_path}")
        return
        
    # Đọc toàn bộ nhãn
    with open(p, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        
    if not lines:
        print("File JSONL trống!")
        return
        
    samples = random.sample(lines, min(num_samples, len(lines)))
    
    fig, axes = plt.subplots(2, 3, figsize=(15, 6))
    axes = axes.flatten()
    
    for i, line in enumerate(samples):
        meta = json.loads(line)
        img_name = meta['file']
        label = meta['label']
        
        img_path = p.parent / img_name
        if not img_path.exists():
            axes[i].set_title("Lỗi: Không thấy ảnh")
            axes[i].axis('off')
            continue
            
        img = cv2.imread(str(img_path))
        if img is None:
            axes[i].set_title("Lỗi: Không đọc được ảnh")
            axes[i].axis('off')
            continue
            
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        axes[i].imshow(img)
        axes[i].set_title(label[:25] + '...' if len(label) > 25 else label, fontsize=14)
        axes[i].axis('off')
        
    plt.tight_layout()
    plt.show()
```

## 1. Dữ liệu Synthetic
Dưới đây là một số mẫu ngẫu nhiên từ tập dữ liệu sinh tự động. Chú ý sự đa dạng về font chữ, màu sắc nền và chữ, độ phân giải.

```python
show_random_samples("../data/synthetic/train/labels.jsonl", num_samples=6)
```

## 2. Dữ liệu ICDAR Recognition (Tiếng Anh)
Dưới đây là các ảnh chữ cắt sẵn thực tế từ các màn hình website/app (ICDAR) đã được xử lý về đúng định dạng JSONL.

```python
show_random_samples("../data/raw/Recognition/labels.jsonl", num_samples=6)
```
